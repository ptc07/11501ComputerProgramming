## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

我們也希望了解：

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 任意精度整數（不易溢位），可依數值大小動態擴充儲存空間 |
| 整數 | `int` | PyLongObject | 與自然數使用同一實作，可表示正數、負數與 0 |
| 實數 | `float` | PyFloatObject | 採 IEEE 754 雙精度浮點數（double）儲存，速度快但有精度誤差 |
| 複數 | `complex` | PyComplexObject | 內含實部與虛部，各以 double 儲存 |
| 集合 | `set` | PySetObject | 基於雜湊表（hash table）實作，查找速度快，元素不可重複 |
| 命題真假 | `bool` | PyBoolObject | bool 是 int 的子類別，只存在 True 與 False 兩個單例物件 |
| 串列 | `list` | PyListObject | 動態陣列（dynamic array），支援索引存取與增減元素 |

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python 的 int 設計目標是:只要記憶體足夠，就要盡可能正確地表示整數
AI 模型設計目標則是:在可接受的誤差範圍內，用最少記憶體完成最多運算
2. 各自的優點與缺點。
整數計算完全精確、幾乎不會整數溢位、適合金融、密碼學、科學計算等需要精確結果的情境
需要更多記憶體、數字愈大運算愈慢、不適合數十億參數的大規模矩陣運算

記憶體需求低、GPU 運算速度快、可訓練數十億到數兆參數模型、提高訓練與推論吞吐量
存在捨入誤差、可能發生數值溢位、可能發生數值下溢、低精度可能影響模型穩定性
3. 為什麼大型語言模型通常選擇較低精度的表示方式？
模型太大、AI 不需要完全精確、GPU 對低精度有硬體加速、實務證明效果足夠好
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
記憶體需求暴增、訓練速度下降、硬體利用率變差、成本增加
